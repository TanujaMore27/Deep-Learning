# Vanishing gradient Problem :
*  in machine learning the vanishing gradient problem is encountered when training ann with gradient-based learning methods and backpropagation.in such methods,during each iteration of training each of the neural networks weights receives an update proportional to the partial derivative of the error function with respect to the current weight.this problem is that in some cases,the gradient will be vanishingly small,effectively preventing the weight from changing its value . in the wrost case,this may completely stop the NN from further training.
* i.e while we are calculating the partial derivative and using the sigmoid function giving output between 0-1 and if their outputs are 0.1 and they are for all in partial derivatives giving so small partial derivative value and when calculating loss the change will be negligiable giving no change in loss i.e loss remaining same.



### How to recognize :
1. focus loss ->epochs->no change->VGP
2. weights graph , take one weight and plot graph of its values against epochs if it remains constant then there is VGP.






### How to reduce Vanishing Gradient Problem :


1. Reduce model complexity :
    we reduce the no. of hidden layers resulting in less multiplication in partial derivatives.but this is not applicable if we are solving too much non-linear problem because we add more layers to solve complex problems then if we reduce the layers then why are we training the model so use this only when when there is not much non-linearity in data.

2.using ReLU activation functions :
   max(0,z) - makes negative zero.but also it has problem of dying ReLU problem that if function becomes zero it will remain zero and there will be no change in the weight while updating.it is solved by leaky ReLU.
