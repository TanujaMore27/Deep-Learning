### What is BackPropagation :
* given the ANN and an error function , the method calculates the gradient of the error function with respect to the neural networks weights.
* It is an algorithm used to train the Neural Networks.
* training means giving best value for weight and bias for the given data.

#### Concept of Gradient :
* just a fancy word for derivative.
* when the function depend on one parameter it is called as derivative.
* when a function is depend on more than one parameter then we use dell instead of 'd' for derivation then it is called as gradient.
* if we are calculating gradient for 9 training parameters it means we are calculating 9 different slopes for w.r.t each dimension.

### Concept of Derivative :
* derivative is the rate of change , if dy/dx = 2 means 1 unit change in x causes 2 unit positive change in y.

*** Derivative at a point :
* at point x this is the slope for dy/dx.

### Concept of Minima :
* for y = x^2 , to find minima we just do dy/dx = 0 , the value of x got is the point of the minima.
* if function is depending upon more than one variables then , we use dell and equal it to 0 and find values for that all variables to get minima.so same is done with loss function.

### Backpropagation Intution :
* we are updating weight and bias such that it will reduce the loss and for each epoch how much that loss will be decreased is same as the slope of that loss function at that point .
* so if dell L/dell W is +ve then we are taking - between Wold and gradient so that loss will decrease and if its -ve then it will becomes -- = + increasing the weight in order to decrease the loss .
* as dell L\ dell W case where its -ve i.e with increasing W loss is decreasing so we update the weight to higher value where minus between old and gradient becomes +. hence, the negative sign between old and gradient for updating values is smartly adjusting the weight and bias in order to reduce the loss.

### Effect of Learning Rate :
* if instead of using lr if we directly taking that values of updating it make the zigzag steps like it will update the parameters in much greater way might where there is no need or limit, so by using learning rate this steps becomes smooth and tiny.
* if learning rate is too low then model will take much time to train and if high then cant get good results so choosing lr values are very crucial.
