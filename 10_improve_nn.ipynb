## How to Improve the Performance of a Neural Network :

1. Fine tuning NN hypermeters:
   * no. of hidden layers:
      	instead of using one hidden layer with too much neurons use multiple hidden layers with few neurons it helps to give best results because DL uses technique called representational learning  in which first layer gives primitive patterns from the data i.e lines , edges,second identifies shapes and last one give the related structure i,e complex patterns.
        include hidden layers till model started overfitting.
   * neurons per layer:
		as for input and output layers nerons are fixed according to classification and features but for hidden layers not necessary to take pyramid patterns because all gives same results . but it should be sufficient enough to capture the patterns within the data i.e if we keep only one neuron in first hidden layer then one one initial pattern is learned and only that is observerd further ignoring remaining ones.
               keep more than suffiecient neurons in layers if it starts to overfit then decrease the neurons till keep it sufficient.
   * learning rate
   * optimizer
   * batch size :
		it has two approaches one is take smaller batch size between 8 to 32 which is slow but gives better results due to generalization.
                second one is large size 8192 depends on GPU which is considered as fast not give better results but can give better results by learning rate schedular if for initial training the learning rate is kept small and after some layers this learning rate is being increased i.e warming up the learning rate which not only makes training faster but also gives better results.
   * activation function
   * no. of epochs :
 		use concept of early stopping which is intelligent enough to understand when should stop. so take any epoch value u can but till early stopping.

2. By solving Problems :

 1.Vanishing / Exploding gradient :

 	* weight initialization
	* change activation function
	* batch normalization
 	*gradient clipping

  2. Not enough data:

	* transfer learning
	* unsupervised pretraining

  3.slow training:

	* optimizers
	* learning rate schedulars

  4. overfitting:
  
	* L1,L2 regularization
	* dropouts
