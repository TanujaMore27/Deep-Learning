# gradient descent in neural network :

* gradient descent is one of the most popular algorithms to perform optimization and by far the most common way to optimize neural networks.
* gradient descent is a way to minimize an objective function J(theta) parameterized by a models parameters Theta belongs to Rd by updating the parameters in the opposite direction of the gradient of the objective function w.r.t the parameters.The learning rate determines the size of the steps we take to reach a minimum.In other words,we follow the direction of the slope of the created by the objective function downhill until we reach a valley.

*** types of gradient descent :
   there are three variants of gradient descent,which differ in how much data we use to compute the gradient of the objective function.Depending on the amount of data, we make a trade-off between the accuracy of the parameter update and the time it takes to perform an update.





1. Batch GD / Vanilla GD:
 * take the whole dataset and calculate gradient descent.
 * will do prediction for all points by dot product getting y_hat array of predictions.then do y-y_hat where having y array of actual outputs to calculate loss.
 * no. of epochs = number of updates
 * so if we have 50 rows updates will calculated 10 times if epochs = 10.
 * because it calculates GD for all in one epoch .

2. Stochastic GB :
* here we take lets epochs = 10 and shuffle the rows in the dataset.
* then again select random points from x array i.e all points from rows taking randomly(to reduce biasness)
* do forward propagation and also calucalte loss and update weights and take averge loss
* so updation will take 50*10 = 500 times
* here updating frequency is high.

*** if batch-size = 1 then it is stochastic GD else it is batch GD

*** Differences :
- Batch GD is faster than Stochastic GD.
- Stochastic is the faster to converge given equal no. of epochs
- loss is stable and smooth in batch while it is unstable and jaggeredness in stochastic GD.
- in stochastic update is made by perspective of each point while in batch it is made by all points of dataset.
- SGD helps the algo to move out of local minima because of its spikey nature while BGD couldn't.
- also SGD may not give exact solution it will give approx. solution.
- in BGD we do vectorization that is calculate dot product for each instead of running in loops making it faster to calculate but due to vectorization we do np.dot(X,W)+b now if X is dataset of 10 crore rows then we first have to load that rows in RAM which is not possible.

3.Mini Batch GD :
- Best of both SBD and BGD
- if we have 320 rows and batch of 32 rows then we have total 10 batches
- so we update 10 times as there are 10 batches.
- so keras have this batch_size = x that divides no of rows with x making batch size of it.
- less spiky curve that of SGD
- vectorization is there but divided into batches doesn't affect RAM.
