### Activation Functions :
* In ANN,each neuron forms a weighted sum of its inputs and passes the resulting scalar value through a function referred to as an activation function or a transfer function.if neuron has n inputs then the output or activation of a neuron is
   a = g(w1x1 + w2x2 + ....+ b) this function g is referred to as an activation function.

#### Why activation functions are needed ?
*  when we take no activation function then the neural network will only perform the linear regression or logistic regression i.e solve only linear problems and not able to solve the non-linear problems.

#### Ideal activation function :
1.Non-Linear : there should be no linear relationship between x and y where y = f(x)
2. Differentiable : to apply gradient descent (not mandatory)
3. computationally inexpensive : easy and fast,not expensive for calculating the derivatives
4.Zero centered : zero centered or normalized so that it will converge fast. i.e mean = 0
5.Non-saturating : does not squeezes inputs within some range.if the function is saturating i.e squeezes inputs within some range then it suffers through vanishing gradient problem.



#### Sigmoid Activation Function :

*** Advantages :
- output within range [0,1]-> used as probability-.binary classification in output layer
- non-linear function -> capture patterns in non linear data
- differentiable -> backpropagation

*** disadvantages :
- saturating function : sufferes through VGP so it is only used in output layers
- non-zero centered : if we apply sigmoid function in all neurons then the values get are not normalized i.e  mean !=0  all numbers are either positive or negative but not zero centered making the training slow,problem in convergence.because while we have only positive and negative values there comes some restrictions while minimizing the loss function so it goes in some zigzag manner requires more time in training.
- computationally expensive due to exponential function.

#### Relu(Rectified Linear Unit) Activation Function :
*** Advantages :
- Non-linear
- not saturated in the positive region
- computationally inexpensive
- convergence is faster

*** Disadvantages :
-  not completely differentiable at zero so for differntiation for x<0 -> 0 and x>=0 -> 1
- not a zero centered -> to solve this we use batch normalization making output normalized
- **Dying Relu Problem :
   	if we use relu in NN then if for any input the result through this activation function is always zero then that becomes the dead neuron,forever remains dead.if more neurons(> 50%) in our network becomes dead then we not able to understand the patterns between our network.
	why ? -> when weighted sum becomes negative due to 1.if our learning rate is high making new weights negative for next cycles. and 2.high negative bias.
	so once because of above reasons weighted some becomes -ve , neurons becomes dead then its not recoverable because only change can be made in inputs and if they are normalized so it will have small value then weighted sum remains -ve all the time and neuron becomes dead.
	solutions : -> 1.set low learning rates
  		       2.bias -> +ve value(0.01)
                       3. dont use relu instead use its variants.
